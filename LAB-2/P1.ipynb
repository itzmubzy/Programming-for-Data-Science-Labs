# Python Basic Structures Review

- String: text data  
  Example: `name = "Rahim"`

- List: ordered, changeable collection  
  Example: `marks = [75, 82, 68, 91]`

- Tuple: ordered, fixed collection  
  Example: `point = (10, 20)`

- Dictionary: key-value data  
  Example: `student = {"name": "Rahim", "marks": 82}`

- For loop: repeats for each item  
  Example: `for mark in marks: ...`

- Function: reusable block of code  
  Example: `def average(values): ...`
