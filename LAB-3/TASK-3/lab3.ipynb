# 3. Selecting and Filtering Data

## Practice — Selecting columns

Use `df["Name"]` for a single column. Use double square brackets for several columns.

```python
# Select one column (show the first five values)
print("Passenger names:")
print(df["Name"].head())

# Select several columns from the dataset
print("\nSelected columns:")
display(df[["Name", "Age", "Sex", "Survived"]].head())
```

## Practice — Selecting rows

- **`loc`** selects rows and columns using their labels.
- **`iloc`** selects rows and columns using integer positions.

In `loc[0:4]`, both endpoints are included. In `iloc[0:5]`, position 5 is excluded.

```python
# Select rows with index labels 0 through 4 and only two columns
print("Selection with loc:")
display(df.loc[0:4, ["Name", "Age"]])

# Select the first five rows and first four columns by position
print("Selection with iloc:")
display(df.iloc[0:5, 0:4])
```

## Problem 2 — Data Filtering

A **filter** keeps only the rows matching a condition. For example, `df["Age"] < 18` checks whether each passenger is younger than 18.

### 1. Display first-class passengers (`Pclass == 1`)

```python
# Keep passengers in first class
first_class = df[df["Pclass"] == 1]

print("Total first-class passengers:", len(first_class))
# Show the first ten matching rows to keep the output readable
display(first_class.head(10))
```

### 2. Find passengers younger than 18

```python
# Keep passengers whose age is below 18
children = df[df["Age"] < 18]

print("Passengers younger than 18:", len(children))
display(children.head(10))
```

### 3. Find male passengers who survived

Use **`&`** to require both conditions to be true. Put each condition inside parentheses.

```python
# Apply two conditions: male AND survived
male_survivors = df[(df["Sex"] == "male") & (df["Survived"] == 1)]

print("Male passengers who survived:", len(male_survivors))
display(male_survivors.head(10))
```

### 4. Find passengers who paid a fare greater than 100

```python
# Keep passengers whose fare is greater than 100
high_fare_passengers = df[df["Fare"] > 100]

print("Passengers who paid more than 100:", len(high_fare_passengers))
display(high_fare_passengers.head(10))
```

**Filtering note:** We show the first **10 matching rows** to keep notebook outputs short. The variables `first_class`, `children`, `male_survivors`, and `high_fare_passengers` contain **all** matching rows. For example, run `display(children)` to view the complete filtered DataFrame.

# 4. Sorting and Aggregation

## Practice — Sorting

`sort_values()` arranges rows based on one or more columns. `ascending=False` means highest first.

```python
# Sort by age: youngest first
print("Youngest passengers:")
display(df.sort_values(by="Age").head(5))

# Sort by fare: highest first
print("Highest fares:")
display(df.sort_values(by="Fare", ascending=False).head(5))

# Sort first by passenger class, then by fare
print("Sorted by passenger class and fare:")
display(df.sort_values(by=["Pclass", "Fare"]).head(5))
```

## Practice — Basic aggregation

Aggregation means summarizing many values into a smaller result.

- `mean()` — average
- `median()` — middle value
- `max()` / `min()` — largest / smallest value
- `value_counts()` — count each category
- `groupby()` — calculate summaries separately for groups

```python
# Calculate a few overall statistics
print("Average age:", round(df["Age"].mean(), 2))
print("Median fare:", round(df["Fare"].median(), 2))
print("Maximum fare:", df["Fare"].max())
print("Minimum fare:", df["Fare"].min())

# Count passengers by gender
print("\nNumber of passengers by gender:")
print(df["Sex"].value_counts())

# Calculate average fares for each passenger class
print("\nAverage fare by class:")
print(df.groupby("Pclass")["Fare"].mean().round(2))

# Since Survived contains 0 or 1, its mean is the survival rate
print("\nSurvival percentage by gender:")
print((df.groupby("Sex")["Survived"].mean() * 100).round(2))
```

## Problem 3 — Sorting and Aggregation

### 1. Display the 10 passengers who paid the highest fares

```python
# Sort fare from highest to lowest and keep ten passengers
top_ten_fares = df.sort_values(by="Fare", ascending=False).head(10)

# Show only the most useful columns
display(top_ten_fares[["Name", "Pclass", "Sex", "Fare"]])
```

### 2. Calculate the average passenger age

```python
# Mean ignores missing values by default
average_age = df["Age"].mean()
print("Average passenger age:", round(average_age, 2), "years")
```

### 3. Count male and female passengers

```python
# Count how often each gender appears
passenger_counts_by_gender = df["Sex"].value_counts()
print(passenger_counts_by_gender)
```

### 4. Calculate the average fare for each passenger class

```python
# Group passengers by class and calculate their average fare
average_fare_by_class = df.groupby("Pclass")["Fare"].mean()
print(average_fare_by_class.round(2))
```

### 5. Calculate the survival rate for each passenger class

`Survived = 1` means survived; `Survived = 0` means did not survive. The **mean** of these 0/1 values is therefore the survival rate. Multiply by `100` to convert to a percentage.

```python
# Calculate survival rate for each class, expressed as a percentage
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
print("Survival rate by passenger class (%):")
print(survival_by_class.round(2))
```

# 5. Final Laboratory Challenge — Titanic Dataset Analysis



### Question 1. How many passengers are in the dataset?

```python
# Each row represents one passenger
number_of_passengers = len(df)
print("Total passengers:", number_of_passengers)
```

### Question 2. What is the average passenger age?

```python
# Calculate the average of available age values
average_passenger_age = df["Age"].mean()
print("Average passenger age:", round(average_passenger_age, 2), "years")
```

### Question 3. Which passenger class has the highest number of passengers?

```python
# Count passengers in each class
class_counts = df["Pclass"].value_counts()
print("Number of passengers in each class:")
print(class_counts.sort_index())

# idxmax() returns the class label with the largest count
most_common_class = class_counts.idxmax()
print("\nClass with the most passengers:", most_common_class)
```

### Question 4. What is the average fare paid by male and female passengers?

```python
# Calculate the mean Fare separately for each gender
average_fare_by_gender = df.groupby("Sex")["Fare"].mean()
print("Average fare by gender:")
print(average_fare_by_gender.round(2))
```

### Question 5. Which gender has a higher survival rate?

```python
# Calculate percentage surviving in each gender group
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100
print("Survival rate by gender (%):")
print(survival_by_gender.round(2))

# Find which gender has the larger survival percentage
higher_survival_gender = survival_by_gender.idxmax()
print("\nGender with higher survival rate:", higher_survival_gender)
```

### Question 6. How does survival rate differ across passenger classes?

```python
# Compare survival percentages across the three passenger classes
class_survival_rates = df.groupby("Pclass")["Survived"].mean() * 100
print("Survival rate by class (%):")
print(class_survival_rates.round(2))

# Identify the class with the highest survival rate
best_survival_class = class_survival_rates.idxmax()
print("\nHighest survival rate was in class:", best_survival_class)
```
